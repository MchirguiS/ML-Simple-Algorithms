```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
training_data = pd.read_csv("Salary_Data.csv")

x_train = training_data["YearsExperience"].values
y_train = training_data["Salary"].values
```

```python
plt.scatter(x_train,y_train)
plt.xlabel("Years of experience")
plt.ylabel("Salary")
```

```python
def cost_function(x, y, w, b):
    cost_sum = 0
    for i in range(len(x)):
        f = w * x[i] + b
        cost_sum += (f - y[i])**2
    return 1 / (2*len(x)) * cost_sum 
```

```python
def gradiant_func(x, y, w, b):
    grad_sum_w = 0
    grad_sum_b = 0
    for i in range(len(x)):
        grad_sum_w += x[i] * ((w * x[i] + b) - y[i])
        grad_sum_b += ((w * x[i] + b) - y[i])
    return (1 / (len(x))) * grad_sum_w, (1 / (len(x))) * grad_sum_b
```

```python
def gradiant_des(x, y, a, iteration):
    w = 0
    b = 0

    for i in range(iteration):
        grad_sum_w, grad_sum_b = gradiant_func(x,y,w,b)

        w = w - a * grad_sum_w
        b = b - a * grad_sum_b

        print(f"iteration {i}:", cost_function(x,y,w,b))

    return w, b
```

```python
alpha = 0.01
iterations = 10000

w, b = gradiant_des(x_train,y_train,alpha,iterations)

print(f"Final w: {w}, Final b: {b}")
```

```python
plt.scatter(x_train,y_train)

x = np.linspace(min(x_train), max(x_train), 100)
y = w * x + b

plt.plot(x,y,color="red",label="Regression Line")
```
